```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
dataset = pd.read_csv("D:/MML/college.csv")
dataset.head(4)
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x="CGPA", y="Package (LPA)", data=dataset)
plt.show()
```

```python
x = dataset[["CGPA"]]
y = dataset[["Package (LPA)"]]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
predicted_package = lr.predict([[7.1]])
predicted_package
```

```python
lr.score(x_test,y_test)
```

```python
# Step 1: Take input from the user
cgpa_input = float(input("Enter CGPA to predict package: "))

# Step 2: Prepare input as DataFrame
new_data = pd.DataFrame([[cgpa_input]], columns=["CGPA"])

# Step 3: Predict using the trained model
predicted_package = lr.predict(new_data)

# Step 4: Display result
print(f"Predicted Package (LPA) for CGPA = {cgpa_input} is: {predicted_package[0][0]:.2f} LPA")
```
