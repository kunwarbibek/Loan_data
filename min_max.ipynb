```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
dataset=pd.read_csv(r"D:\MML\loan_data.csv")
dataset.head(5)
```

```python
dataset.isnull().sum()
```

```python
dataset.describe()
```

```python
from sklearn.preprocessing import MinMaxScaler
ms = MinMaxScaler()
ms.fit(dataset[["CoapplicantIncome"]])
dataset['CoapplicantIncome_minmax']=ms.transform(dataset[["CoapplicantIncome"]])
dataset.head(3)
```

```python

plt.figure(figsize=(12,5))
plt.subplot(1, 2, 1)
plt.title("Before ")
sns.displot(dataset["CoapplicantIncome"])


plt.figure(figsize=(12,5))
plt.subplot(1, 2, 1)
plt.title("After ")
sns.displot(dataset["CoapplicantIncome_minmax"])
plt.show()
```
