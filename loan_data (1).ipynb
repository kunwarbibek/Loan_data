```python
# Step 4: Handle missing values
# 1. Fill missing categorical columns with mode
categorical_cols = ['Gender', 'Married', 'Dependents', 'Self_Employed']
for col in categorical_cols:
    dataset[col] = dataset[col].fillna(dataset[col].mode()[0])
```

```python
dataset.head(4)
```

```python
#numricals_cols= ['LoanAmount', 'Loan_Amount_Term', 'Credit_History', 'ApplicantIncome', 'CoapplicantIncome']
#for col in numricals_cols:
   # dataset[col]=dataset[col].fillna(dataset[col].mean()
# method for inserting numrical missing value
numerical_cols = ['LoanAmount', 'Loan_Amount_Term', 'Credit_History', 'ApplicantIncome', 'CoapplicantIncome']
for col in numerical_cols:
    dataset[col] = dataset[col].fillna(dataset[col].mean())

dataset.head(5)

```

```python
dataset.isnull().sum()
```

```python
#corrected data types 
dataset['Dependents']=dataset['Dependents'].replace('3+', 3).astype(float)
dataset['Credit_History'] = dataset['Credit_History'].astype(int)
dataset['Loan_Amount_Term'] = dataset['Loan_Amount_Term'].astype(int)
dataset['LoanAmount'] = dataset['LoanAmount'].astype(float)
```

```python
dataset['Gender'] = dataset['Gender'].map({'Male': 'Male', 'Female': 'Female'})
dataset['Married'] = dataset['Married'].map({'Yes': 1, 'No': 0})
dataset['Education'] = dataset['Education'].map({'Graduate': 1, 'Not Graduate': 0})
dataset['Self_Employed'] = dataset['Self_Employed'].map({'Yes': 1, 'No': 0})
dataset['Loan_Status'] = dataset['Loan_Status'].map({'Y': 1, 'N': 0})
```

```python
# Step 7: Remove duplicates if any
dataset.drop_duplicates(inplace=True)
```
